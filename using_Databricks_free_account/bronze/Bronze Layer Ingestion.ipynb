```python
SOURCES=[{
    "source":"crm",
    "path":"/Volumes/dwh_project/bronze/data_sets/source_crm/cust_info.csv",
    "file":"cust_info"
},{
    "source":"crm",
    "path":"/Volumes/dwh_project/bronze/data_sets/source_crm/prd_info.csv",
    "file":"prd_info"
},{
    "source":"crm",
    "path":"/Volumes/dwh_project/bronze/data_sets/source_crm/sales_details.csv",
    "file":"sales_details"
},{
    "source":"erp",
    "path":"/Volumes/dwh_project/bronze/data_sets/source_erp/CUST_AZ12.csv",
    "file":"CUST_AZ12"
},{
    "source":"erp",
    "path":"/Volumes/dwh_project/bronze/data_sets/source_erp/LOC_A101.csv",
    "file":"LOC_A101"
},{
    "source":"erp",
    "path":"/Volumes/dwh_project/bronze/data_sets/source_erp/PX_CAT_G1V2.csv",
    "file":"PX_CAT_G1V2"
}
]
```

```python
for source in SOURCES:

    print(f"loading file {source["file"]} from {source["path"]}")
    df=(
        spark.read.format("csv")
        .option("header","true")
        .option("inferSchema","true")
        .load(source["path"])
    )

    df.write.mode("overwrite").format("delta").saveAsTable(f"dwh_project.bronze.{source['file']}")
```
