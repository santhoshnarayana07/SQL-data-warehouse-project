```python
import pyspark.sql.functions as f
df=spark.read.table("dwh_project.bronze.cust_az12")
df=(df.withColumn("CID",
                  f.when(f.length("CID")==13,f.regexp_replace(f.col("CID"),"NAS",""))
                  .otherwise(f.col("CID"))
                  )
    )
```

```python
import pyspark.sql.functions as f
df=df.withColumn("GEN",
                f.when(f.upper(f.trim(f.col("GEN"))).isin("MALE","M"),"male")
                .when(f.upper(f.trim(f.col("GEN"))).isin("FEMALE","F"),"female")
                .otherwise("n/a")
                )
df= df.filter(f.col("BDATE")<f.to_date(f.current_date()))
```

```python
df.write.mode("overwrite").saveAsTable("dwh_project.silver.cust_az12")
```
