```python
import pyspark.sql.functions as f
df=(spark.read.table("dwh_project.bronze.loc_a101"))

df=(
    df.withColumn("CNTRY",f.when((f.trim(f.col("CNTRY"))=="")|(f.trim(f.col("CNTRY")).isNull()) ,"n/a")
                  .when(f.upper(f.trim(f.col("CNTRY"))).isin("US","UNITED STATES","USA"),"United States")
                  .when(f.upper(f.trim(f.col("CNTRY")))=="DE","Germany")
                  .otherwise(f.col("CNTRY"))
                )
    .withColumn("CID",f.regexp_replace("CID","-",""))
)

```

```python
df.write.mode("overwrite").saveAsTable("dwh_project.silver.loc_a101")
```
