```python
%sql
DROP VIEW IF EXISTS dwh_project.gold.dim_prd;
CREATE VIEW dwh_project.gold.dim_prd as
select 
row_number() over(order by prd_id asc) as ID,
prd_id as product_id,
cat_id as category ,
cat.SUBCAT as sub_category,
prd_key as product_key,
prd_nm as product_name,
prd_cost as product_cost,
cat.MAINTENANCE as maintenanct,
prd_line ,
prd_start_dt as starting_date
from dwh_project.silver.prd_info as prd
left join dwh_project.silver.px_cat_g1v2 as cat
on cat.ID=prd.cat_id
```
