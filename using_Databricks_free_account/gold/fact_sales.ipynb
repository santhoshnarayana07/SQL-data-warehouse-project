```python
%sql
DROP VIEW IF EXISTS dwh_project.gold.fact_sales;
create view dwh_project.gold.fact_sales as
select 
sls_ord_num as order_id,
prd.ID as product_id,
cst.ID as cutomer_id,
sls_order_dt as order_date,
sls_ship_dt as shipping_date,
sls_due_dt as due_date,
sls_sales as sales,
sls_quantity as quantity,
sls_price as price
from dwh_project.silver.sales_details as sls
left join dwh_project.gold.dim_prd as prd
on prd.product_key=sls.sls_prd_key
left join dwh_project.gold.dim_sales as cst
on cst.customer_id=sls.sls_cust_id
```
