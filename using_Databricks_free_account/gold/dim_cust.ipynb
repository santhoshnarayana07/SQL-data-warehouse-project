```python
%sql

DROP VIEW IF EXISTS dwh_project.gold.dim_sales;

CREATE VIEW dwh_project.gold.dim_sales as 
select 
row_number()over(order by cst_id asc) as ID,
cst_id as customer_id,
cst_key as customer_key,
cst_firstname as first_name,
cst_lastname as last_name,
gen.BDATE as birth_date,
CASE 
    WHEN trim(gen.GEN)="n/a" or gen.GEN IS NULL THEN COALESCE(cst.cst_gndr,"n/a")
    ELSE gen.GEN
END AS gender,
cst_marital_status as marital_status ,
cnt.CNTRY
from dwh_project.silver.cust_info as cst
left join dwh_project.silver.cust_az12 as gen
ON cst.cst_key=gen.CID
left join dwh_project.silver.loc_a101 as cnt
on cnt.CID=cst.cst_key
```
